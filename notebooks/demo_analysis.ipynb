# Advanced Climate Modeling Platform — Demo Notebook

This notebook demonstrates interactive exploration of the climate platform outputs.

**Prerequisites:** Run `python -m src.pipeline` first.

```python
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load processed data
ds = xr.open_zarr('../data/zarr/processed_climate.zarr')
print(ds)
```

```python
# Global mean temperature
df = pd.read_csv('../outputs/reports/global_means.csv', parse_dates=['time'])
df.plot(x='time', y='tas_global_K', title='Global Mean Temperature', figsize=(10, 4))
plt.tight_layout()
plt.show()
```

```python
# Ensemble spread
ens = pd.read_csv('../outputs/reports/ensemble_stats.csv', index_col='time', parse_dates=True)
member_cols = [c for c in ens.columns if c.startswith('member_')]

fig, ax = plt.subplots(figsize=(10, 4))
for col in member_cols:
    ax.plot(ens.index, ens[col], alpha=0.5, linewidth=1)
ax.plot(ens.index, ens['ensemble_mean'], 'k-', linewidth=2, label='Ensemble Mean')
ax.set_title('Ensemble Temperature')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Spatial temperature map
tas_first = ds['tas'].isel(time=0).compute()

fig, ax = plt.subplots(figsize=(12, 5))
im = ax.pcolormesh(ds['lon'].values, ds['lat'].values, tas_first.values,
                   cmap='RdBu_r', shading='auto')
plt.colorbar(im, ax=ax, label='Temperature (K)')
ax.set_title('Surface Temperature — January 2000')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
plt.tight_layout()
plt.show()
```
